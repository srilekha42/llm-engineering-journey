```python
# 1. Install prerequisite library for audio file handling
!pip install -q soundfile

import torch
from transformers import pipeline
import soundfile as sf
from google.colab import userdata
from huggingface_hub import login

# 2. Login using your vault key
hf_token = userdata.get('HF_TOKEN')
login(token=hf_token, add_to_git_credential=False)

# 3. Load the Text-to-Speech model into VRAM on the GPU (device=0)
print("Loading model into VRAM...")
tts_pipe = pipeline("text-to-speech", model="microsoft/speecht5_tts", device=0)

# 4. Create a clean vector profile directly in memory
print("Creating vocal profile vector...")
speaker_embedding = torch.zeros((1, 512)).to(0)

# 5. Run the inference math directly on the graphics hardware
text_prompt = "System initialized. Engineer, you are running locally on a cloud graphics cluster."
print("Generating audio array on GPU...")
speech = tts_pipe(text_prompt, forward_params={"speaker_embeddings": speaker_embedding})

# 6. Output the final raw array as a playable wav file
output_filename = "mastery_speech.wav"
sf.write(output_filename, speech["audio"], speech["sampling_rate"])
print(f"\nSuccess! Audio file compiled and saved as: {output_filename}")
```

```python
# 1. Install the prerequisite library for handling open-source diffusion models
!pip install -q diffusers transformers accelerate

import torch
from diffusers import AutoPipelineForText2Image

# 2. Initialize the Turbo diffusion pipeline straight into the T4 GPU VRAM
print("Loading 5GB SDXL Turbo model into VRAM (this takes a couple of minutes to download)...")
pipe = AutoPipelineForText2Image.from_pretrained(
    "stabilityai/sdxl-turbo",
    torch_dtype=torch.float16,
    variant="fp16"
).to("cuda")

# 3. Define the pop-art prompt
prompt = "A class of data scientists learning AI engineering in a vibrant, high-energy pop art style."

# 4. Run the inference loop on the GPU (Only 1 step needed for Turbo!)
print("GPU is processing matrix math to generate your image...")
image = pipe(prompt=prompt, num_inference_steps=1, guidance_scale=0.0).images[0]

# 5. Export the raw tensor pixel data as a standard image file
output_image_path = "pop_art_classroom.png"
image.save(output_image_path)
print(f"\nSuccess! Image rendered and saved as: {output_image_path}")
```
